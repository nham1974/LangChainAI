#### Most models can stream their output content while it is being generated. By displaying o/p progressively , streaming significantly improves user experience , particularly for longer responses.
#### Calling Stream() return iterator that yields o/p chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("why do parrots talk ?")
response.content
## stream()
model.stream("write me a 200 words paragraph on Artificail intelligence")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers"):
    print(chunk.text,end="|",flush=True)
```

#### Batch : Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the priocessing can be done in parallel.

```python
responses = model.batch([
    "why do parrots have colored feathers ?",
    "why do airplanes fly?",
    "what is quantun computing?"]
)
for response in responses:
    print(response)
```

```python
responses = model.batch([
    "why do parrots have colored feathers ?",
    "why do airplanes fly?",
    "what is quantun computing?"],
    config={
        'max_concurrency':5, # limit to 5 parallel calls
    }
)
for response in responses:
    print(response)
```

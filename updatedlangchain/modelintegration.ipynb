#### Model Integrations with OpenAI, groq and google gemini

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")
model
```

```python
## invoke model
response = model.invoke("Hello! how are you ?")
#response
response.content
```

```python
## Google gemini integration
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
googlemodel = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = googlemodel.invoke("why do parrots talk ?")
response.content
```

```python
## integrate with Groq
import os
from langchain.chat_models import init_chat_model
from langchain_groq import ChatGroq # another way of using groq

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("why do parrots talk ?")
response
```
